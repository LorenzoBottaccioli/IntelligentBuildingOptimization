# Lab: from an IDF to a surrogate-based design optimisation


In this laboratory you will go over the basic workflow to create a surrogate design-optimisation model from an EnergyPlus simulation. You will train surrogate models and use the best one to find optimal design parameters.

The toolbox is **bdopt**. The objects you need, step by step, are listed in the next cell; the cells below are empty on purpose: this is your notebook.

| Step | bdopt |
|---|---|
| read the model | `load_idf` |
| where a value goes in the model | `FieldSelector` (`object_name='*'` for every object of the class, a list of names, or `where=` to keep some of them); `MultiSelector.from_targets(names, class_name=, field_name=)` for one value written in many objects |
| which values are allowed | `RangeParameter`, `IntegerParameter`, `CategoryParameter` |
| a design variable | `Parameter(selector, value_descriptor, name)`: one `Parameter` is one column |
| the problem | `Problem(parameters, objectives)`; the objectives are EnergyPlus meters, e.g. `'Electricity:Facility'` |
| samples | `sampling.lhs(problem, n)` |
| simulate | `EnergyPlusEvaluator(problem, idf, epw_file, ...)`, then `evaluator.evaluate(samples, keep_input=True)`; the model asks for 38 `Output:Variable` objects that the objectives do not need: simulate a copy without them (`idf.copy()`, `remove(by_class('Output:Variable'))`) |
| plots | `viz.pairplot`, `viz.correlation_heatmap`, `viz.parallel_coordinates`, `viz.pareto` |
| surrogates | `GPSurrogate`, `MLPSurrogate`, `AutoSurrogate` (`fit(table, problem)`, `predict`, `score`), or scikit-learn |
| a function instead of EnergyPlus | `FunctionEvaluator(problem, function, vectorized=True)` |
| optimisation | `nsga2(problem, model)`, then `front.df` and `front.closest_to_ideal()` |
| write a design into the model | `problem.apply(idf, design, idd)` and `idf.save('file.idf')` |
